## Python Memory Management
Memory management in Python involves a combination of automatic garbage collection, reference counting, and various internal optimizations to effiecintly manage memory allocation and deallocation. understanding these mechanisms can help developers wirte more effiecient and robust application. 

```python
### Reference Counting
# Python uses reference counting as one of its memory management techniques. Each object in Python has a reference count that keeps track of how many references point to that object. When the reference count drops to zero, the memory occupied by the object is deallocated.
```

```python
import sys

a = []
print(sys.getrefcount(a))  # Output: 2 (one for 'a' and one for the argument passed to getrefcount)
```

```python
b = a
print(sys.getrefcount(a))  # Output: 3 (one for 'a', one for 'b', and one for the argument passed to getrefcount)
```

```python
del b # Deleting 'b' will decrease the reference count of the list object
print(sys.getrefcount(a))  # Output: 2 (one for 'a' and one for the argument passed to getrefcount)
```

```python
### Garbage Collection
# In addition to reference counting, Python also has a garbage collection mechanism to handle cyclic references.
# Python's garbage collector uses a generational approach, where objects are categorized into three generations based on their age. The garbage collector periodically runs to identify and collect objects that are no longer in use, especially those involved in reference cycles.
```

```python
import gc
gc.enable()  # Enable garbage collection
```

```python
gc.disable() # Disable garbage collection if needed
```

```python
gc.collect() # Manually trigger garbage collection
```

```python
### Get garbage collection stats
print(gc.get_stats())
```

```python
### Get Unreachable objects
print(gc.garbage) # empty because there are no unreachable objects
```

### Memory Management Best Practices
1. Use Local vairables: Local variable have a shorter lifespan and are freed sooner than global variables.
2. Avoid circular references: Be cautious when creating objects that reference each other, as this can lead to memory leaks if not handled properly.
3. Use Generators: Generators produce items one at a time and only keep one item in memory, which can help reduce memory usage for large datasets.
4. Explicitly delete objects: Use the `del` statement to explicitly delete objects that are no longer needed, which can help free up memory sooner.
5. Profile memory usage: Use tools like `memory_profiler` to identify memory usage patterns and optimize your code accordingly.

```python
import gc

class Myobject:
    def __init__(self,name):
        self.name = name
        print(f"Object {self.name} created")

    def __del__(self):
        print(f"Object {self.name} destroyed")

## Create circluar reference
obj1 = Myobject("Object1")
obj2 = Myobject("Object2")
obj1.ref = obj2
obj2.ref = obj1

del obj1
del obj2
# If we run this code, we will see that the __del__ method is not called for either object, indicating that they are not being garbage collected due to the circular reference.


## Manually trigger garbage collection to see if the objects are collected
gc.collect() # This will not call the __del__ method for either object due to the circular reference, and they will be added to gc.garbage if they are not collected.


        
```

```python
### Generators for Memory Efficiency
# Generators are a powerful tool in Python for creating iterators that can produce items one at a time, which can be more memory efficient than creating a list of all items at once.

def generate_numbers(n):
    for i in range(n):
        yield i

# Using the generator
for num in generate_numbers(10):
    print(num)
    if num == 5:
        break  # We can stop the generator early without generating all numbers up to n, saving memory.
```

```python
### Profiling memory usage
import tracemalloc

def create_large_list(n):
    return [i for i in range(10000)]

def main():
    tracemalloc.start()  # Start tracing memory allocations
    create_large_list(10000)

    snapshot = tracemalloc.take_snapshot()  # Take a snapshot of memory usage
    top_stats = snapshot.statistics('lineno')  # Get statistics by line number

    print("[ Top 10 ]")
    for stat in top_stats[:10]:
        print(stat)
```

```python
main()
```
